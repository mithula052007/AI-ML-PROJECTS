```python
sns.barplot(x="school",
           y="absences",
           data=df)
plt.title("absences in school")
plt.show()
```

```python
line_data = df.groupby("school")["absences"].sum()
plt.plot(line_data.index,
        line_data.values,
        marker="o")
plt.title("absences in school")
plt.xlabel("school")
plt.ylabel("absences")
plt.show()
```

```python
sns.histplot(df["absences"], kde=True)
plt.title("absences in school")
plt.show()
```
