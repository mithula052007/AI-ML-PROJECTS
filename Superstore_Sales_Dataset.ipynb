```python
sns.barplot(x="Category",
           y="Sales",
           data=df)
plt.title("Sales by Category")
plt.show()
```

```python
line_data = df.groupby("Region")["Sales"].sum()
plt.plot(line_data.index,
        line_data.values,
        marker="o")
plt.title("Region-wise Sales")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
sns.histplot(df["Sales"], kde=True)
plt.title("Sales Distribution")
plt.show()
```

```python
sns.scatterplot(x="Region",
               y="Sales",
               data=df)
plt.title("Sales vs Profit")
plt.show()
```
